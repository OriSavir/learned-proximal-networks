```python
import numpy as np
import matplotlib.pyplot as plt
import json
import os
```

```python
with open('./optimal_scale_NE_64.json') as f:
    data = json.load(f)
    ne_optimal_scales = {float(k): v for k, v in data.items()}
print(ne_optimal_scales)
```

```python
with open('./optimal_scale_LPN.json') as f:
    data = json.load(f)
    lpn_optimal_scales = {float(k): v for k, v in data.items()}
print(lpn_optimal_scales)
```

```python
save_dir = './test_result_figures/'
os.makedirs(save_dir, exist_ok=True)
NE_data_directory = './ne_lpn_results/inverse/deblur/' 
LPN_data_directory = './original_lpn_results/inverse/deblur/'
```

```python
sigma_noise = [0.02, 0.05, 0.1, 0.2, 0.3, 0.5]
```

```python
for noise_level in sigma_noise:
    NE_blur = 2.0
    LPN_blur = 2.0
    NE_path = f'{NE_data_directory}blur={NE_blur}_noise={noise_level}/admm/lpn'
    LPN_path = f'{LPN_data_directory}blur={LPN_blur}_noise={noise_level}/admm/lpn' 

    ground_truth1 = np.load(f'{NE_path}/x/0.npy')

    NE_noised1 = np.load(f'{NE_path}/y/0.npy')
    NE_deblurred1 = np.load(f'{NE_path}/xhat/0.npy')

    LPN_noised1 = np.load(f'{LPN_path}/y/0.npy')
    LPN_deblurred1 = np.load(f'{LPN_path}/xhat/0.npy')

    ground_truth2 = np.load(f'{NE_path}/x/1.npy')

    NE_noised2 = np.load(f'{NE_path}/y/1.npy')
    NE_deblurred2 = np.load(f'{NE_path}/xhat/1.npy')

    LPN_noised2 = np.load(f'{LPN_path}/y/1.npy')
    LPN_deblurred2 = np.load(f'{LPN_path}/xhat/1.npy')

    # My notes for plotting here: Five Columns: Ground Truth, NE Blurred, NE Deblurred, LPN Blurred, LPN Deblurred. Two rows: Noised1, Noised2
    # Below NE section (the two NE columns), put the NE blur value, and below the LPN section, put the LPN blur value
    # Title: Noise Level: {noise_level}

    fig, axes = plt.subplots(2, 5, figsize=(15, 6))

    col_titles = ["Ground Truth", "NE Blurred", "NE Deblurred", "LPN Blurred", "LPN Deblurred"]
    
    images_row1 = [ground_truth1, NE_noised1, NE_deblurred1, LPN_noised1, LPN_deblurred1]
    images_row2 = [ground_truth2, NE_noised2, NE_deblurred2, LPN_noised2, LPN_deblurred2]

    for i, (img, title) in enumerate(zip(images_row1, col_titles)):
        axes[0, i].imshow(img)
        axes[0, i].axis('off')
        axes[0, i].set_title(title, fontsize=10)

    for i, img in enumerate(images_row2):
        axes[1, i].imshow(img)
        axes[1, i].axis('off')
    
    NE_scale = ne_optimal_scales[noise_level]
    LPN_scale = lpn_optimal_scales[noise_level]

    fig.text(0.25, 0.02, r"$\sigma_{\mathrm{blur, NE}}$ = " + str(NE_blur), fontsize=18)
    fig.text(0.65, 0.02, r"$\sigma_{\mathrm{blur, LPN}}$ = " + str(LPN_blur), fontsize=18)

    plt.suptitle(r"$\sigma_{\mathrm{noise}}$ = " + str(noise_level), fontsize=18, fontweight='bold')

    plt.tight_layout(rect=[0, 0.05, 1, 0.95])
    plt.subplots_adjust(top=0.85) 

    filename = f"noise={noise_level}_NEblur={NE_blur}_LPNblur={LPN_blur}.png"
    save_path = os.path.join(save_dir, filename)
    plt.savefig(save_path, dpi=300)
    print(f"Saved: {save_path}")

    plt.close(fig)  
    
```
